## Step 3: Missing Values Audit

Missing values can affect data analysis, reporting, and machine learning models.

In this step, we will:
- Identify columns containing missing values.
- Count the number of missing values.
- Calculate the percentage of missing values.
- Create a summary table for the audit.

A column with a high percentage of missing values requires further investigation before analysis.

```python
# Count missing values in each column

missing_count = df.isnull().sum()

missing_count
```

### 3.1 Missing Value Percentage

The percentage of missing values is calculated to understand the severity of missing data in each column.

```python
# Calculate missing value percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage
```

### 3.2 Create Missing Value Audit Table

A consolidated audit table is created containing:
- Column name
- Missing value count
- Missing value percentage

Only columns with missing values are displayed.

```python
# Create missing value audit table

missing_audit = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isnull().sum().values,
    "Missing_Percentage": missing_percentage.values
})

# Show only columns with missing values
missing_audit = missing_audit[
    missing_audit["Missing_Count"] > 0
].sort_values(
    by="Missing_Count",
    ascending=False
)

missing_audit
```

### 3.3 Missing Value Audit Summary

The audit table above identifies the columns affected by missing values.

If no rows are returned, the dataset contains no missing values.

If missing values are present, they will be recorded in the Issue Log and investigated before deciding whether to remove, replace, or retain them.

```python
# Overall missing value check

total_missing = df.isnull().sum().sum()

print("Total Missing Values:", total_missing)

if total_missing == 0:
    print("Result: No missing values found.")
else:
    print("Result: Missing values found. Further investigation is required.")
```

### Audit Finding

The missing-value check has been completed.

The total number of missing values and the affected columns have been quantified. These findings will be used later to prepare the Data Quality Issue Log.

### 5.1 Validate Sales

Sales represents the monetary value of an order.

```python
# Check invalid Sales values

invalid_sales = df[df["Sales"] < 0]

print("Invalid Sales Records:", len(invalid_sales))
```

Quantity represents the number of units sold.

Rule:
- Quantity should be greater than 0.

```python
# Check invalid Quantity values

invalid_quantity = df[df["Quantity"] <= 0]

print("Invalid Quantity Records:", len(invalid_quantity))
```

For example:
- 0 = 0% discount

## Step 7: Categorical Consistency Audit

Categorical values should be consistent throughout the dataset.

In this step, we will inspect important categorical columns for:
- Unexpected values
- Different spellings
- Extra spaces
- Inconsistent categories

The following columns will be checked:
- Segment
- Category
- Region
- Ship Mode

### 7.1 Check Unique Values

The `unique()` function is used to identify all distinct values in important categorical columns.

```python
# Check unique values in categorical columns

categorical_columns = [
    "Segment",
    "Category",
    "Region",
    "Ship Mode"
]

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

### 7.2 Check Frequency of Categorical Values

Value counts are used to understand the frequency of each category and identify unusual or unexpected values.

```python
# Display frequency of categorical values

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts())
```

### 7.3 Check for Leading or Trailing Spaces

Leading or trailing spaces can create duplicate categories that appear different but actually represent the same value.

We will check text columns for unnecessary spaces.

```python
# Check text columns for leading/trailing spaces

text_columns = df.select_dtypes(include="object").columns

space_issues = {}

for column in text_columns:
    count = (
        df[column].astype(str)
        .str.strip()
        .ne(df[column].astype(str))
        .sum()
    )
    
    if count > 0:
        space_issues[column] = count

space_issues
```

### 7.5 Categorical Consistency Finding

The categorical columns were checked for unexpected values and leading or trailing spaces.

The audit found no leading or trailing space issues in the text columns.

The main categorical fields also contain consistent category values.

Therefore, no categorical standardization was required at this stage.

## Step 8: Data Quality Issue Log

A Data Quality Issue Log summarizes the results of all validation checks performed during the audit.

The issue log contains:
- Data quality check
- Validation rule
- Number of affected records
- Status
- Action required

This provides a repeatable and auditable record of the dataset quality.

```python
# Create Data Quality Issue Log

issue_log = pd.DataFrame({
    "Check": [
        "Missing Values",
        "Duplicate Records",
        "Sales Range",
        "Quantity Range",
        "Discount Range",
        "Order Date Validity",
        "Ship Date Validity",
        "Categorical Consistency"
    ],
    "Validation Rule": [
        "No missing values",
        "No duplicate rows",
        "Sales >= 0",
        "Quantity > 0",
        "Discount between 0 and 1",
        "Valid datetime",
        "Valid datetime",
        "Consistent categorical values"
    ],
    "Affected_Records": [
        total_missing,
        duplicate_count,
        len(invalid_sales),
        len(invalid_quantity),
        len(invalid_discount),
        invalid_order_dates,
        invalid_ship_dates,
        sum(space_issues.values())
    ]
})

issue_log["Status"] = issue_log["Affected_Records"].apply(
    lambda x: "PASS" if x == 0 else "ISSUE FOUND"
)

issue_log
```

### 8.1 Issue Log Interpretation

A status of PASS indicates that no records violated the corresponding validation rule.

An ISSUE FOUND status indicates that further investigation or corrective action is required.

The issue log will be included in the final Data Quality Audit Report.

```python
# Count passed and failed checks

print("Total Checks:", len(issue_log))
print("Passed Checks:", (issue_log["Status"] == "PASS").sum())
print("Issues Found:", (issue_log["Status"] == "ISSUE FOUND").sum())
```
