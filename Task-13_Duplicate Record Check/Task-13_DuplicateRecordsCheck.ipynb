```python
import pandas as pd
df = pd.read_excel("../data/raw/sample_-_superstore.xls")
```

```python
df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Check complete duplicate rows
duplicate_count = df.duplicated().sum()

print("Complete duplicate rows:", duplicate_count)
```

```python
duplicate_rows = df[df.duplicated(keep=False)]

print("Duplicate rows found:")
display(duplicate_rows)
```

```python
# Check duplicate Row IDs
duplicate_row_ids = df["Row ID"].duplicated().sum()

print("Duplicate Row IDs:", duplicate_row_ids)
```

```python
# Check records belonging to repeated Order IDs
repeated_order_records = df[
    df["Order ID"].duplicated(keep=False)
]

print("Records with repeated Order IDs:",
      len(repeated_order_records))
```

```python
repeated_order_records[
    ["Row ID", "Order ID", "Product Name", "Sales", "Quantity", "Profit"]
].head(10)
```

```python
# Check repeated Order ID + Product Name combinations

duplicate_order_product = df[
    df.duplicated(
        subset=["Order ID", "Product Name"],
        keep=False
    )
]

print(
    "Records with repeated Order ID + Product Name:",
    len(duplicate_order_product)
)
```

```python
duplicate_order_product[
    ["Row ID", "Order ID", "Product Name",
     "Sales", "Quantity", "Discount", "Profit"]
]
```

```python
duplicate_summary = pd.DataFrame({
    "Check": [
        "Total Records",
        "Exact Duplicate Rows",
        "Duplicate Row IDs",
        "Records with Repeated Order IDs",
        "Records with Repeated Order ID + Product"
    ],
    "Result": [
        len(df),
        df.duplicated().sum(),
        df["Row ID"].duplicated().sum(),
        df["Order ID"].duplicated(keep=False).sum(),
        df.duplicated(
            subset=["Order ID", "Product Name"],
            keep=False
        ).sum()
    ]
})

display(duplicate_summary)
```

```python
# Remove only complete duplicate rows
df_cleaned = df.drop_duplicates()

print("Original rows:", len(df))
print("Cleaned rows:", len(df_cleaned))
print("Rows removed:", len(df) - len(df_cleaned))
```

```python
print("Duplicate rows after cleaning:",
      df_cleaned.duplicated().sum())

print("Final shape:",
      df_cleaned.shape)
```

```python
output_file = "Task-13_Duplicate_Check_Cleaned.csv"

df_cleaned.to_csv(output_file, index=False)

print("Cleaned dataset saved as:", output_file)
```

```python
duplicate_summary.to_csv(
    "Duplicate_Report.csv",
    index=False
)

print("Duplicate report saved.")
```
